```python
from transformers import pipeline
```

```python
pipe = pipeline("table-question-answering", model="google/tapas-base-finetuned-wtq")
```

```python
import pandas as pd
```

```python
table = pd.read_csv("/content/YOUTUBE CHANNELS DATASET.csv")
```

```python
def get_answer(table, question):
  table = table.astype(str)
  return pipe(table=table, query=question)
```

```python
get_answer(table, "who has highest subscribers?")
```

```python
get_answer(table, "T-series has how many subscribers?")
```

```python
pipe2 = pipeline("table-question-answering", model="google/tapas-large-finetuned-wtq")
```

```python
def get_answer2(table, question):
  table = table.astype(str)
  return pipe2(table=table, query=question)
```

```python
get_answer2(table, "T-series has how many subscribers?")
```

```python
get_answer2(table, "Who has most subscribers WWE or PewDiePie?")
```

```python
get_answer2(table, "Like Nastya views?")
```
